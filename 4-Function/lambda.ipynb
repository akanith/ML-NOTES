### Lambda Function Syntax
Anatomy of a single-line anonymous function.


```python
##Syntax 

lambda arguments : expression
```

### Named Function Example
Standard function definition for addition.


```python
def addition(a,b):
    return a+b
```

### Calling the Named Function


```python
addition(2,3)
```

### Lambda Function for Addition
Equivalent logic using a lambda expression.


```python
addition = lambda a,b:a+b
print(addition(10,5))
```

### Named Function for Parity
Standard function for checking even numbers.


```python
def even(num):
    if num%2 == 0:
        return True

even(24)
```

### Lambda Function for Parity
Concise even-check using lambda.


```python
even1=lambda num:num%2==0
even1(10)
```

### Multiple Arguments in Functions
Comparing standard and lambda functions with 3 arguments.


```python
def addition(x,y,z):
    return x+y+z

addition1=lambda x,y,z:x+y+z
addition1(2,5,3)
```

### Procedural Even/Odd Check
Logic implemented without a function.


```python
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

### Functional Even/Odd Check
Encapsulating logic for reuse.


```python
def even_or_odd(num):
    """This function is for Even or Odd"""
    if num%2==0:
        print("Even")
    else:
        print("Odd")

num=int(input("Enter a Number"))
even_or_odd(num)
```

### Multiple Parameters and Return

```python
## Function with multiple parameters

def add(a,b):
    c=a+b
    return c

result=add(10,15)
print(result)
```

```python
## Defailt Parameters

def greet(name="Gust"):
    print(f"Hello {name},Good Morning")

greet("Anith")
```

### Positional Variable-Length Arguments (*args)
Handling an arbitrary number of positional inputs.


```python
## Variable Length Arguments 
## Postional And Keywords arugment 

def print_function(*args):
    for number in args:
        print(number)
```

### Calling Functions with *args


```python
print_function(1,2,3,4,5,6,7,8,9,"Anith")
```

### Keyword Variable-Length Arguments (**kwargs)
Handling an arbitrary number of named inputs.


```python
def print_details(**kwards):
    for key,value in kwards.items():
        print(f"{key}:{value}")
        
```

### Calling Functions with **kwargs


```python
print_details(name="Anith",age="20",country="India")
```

### Combining *args and **kwargs
Managing both positional and keyword variable arguments.


```python
def print_details(*args,**kwards):
    for val in args:
        print(f"Positional Aurgment : {val}")

    for key,value in kwards.items():
        print(f"{key}:{value}")
```

### Comprehensive Function Call Example


```python
print_details(1,2,3,4,"Anith",name="Anith",age="20",country="India")
```

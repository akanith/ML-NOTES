### Introduction to Pandas Series
Creating 1D labeled arrays from lists.


```python
##Series

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("series \n",series)
```

### Creating Series from Dictionaries
Automatic index mapping from keys.


```python
##create a series from dictonary

data={'a':1,'b':2,'c':3}
ser_dict=pd.Series(data)
print(ser_dict)
```

### Custom Indexing in Series
Explicitly defining labels for series elements.


```python
data=[1,2,3]
index=['a','b','c']
pd.Series(data,index=index)
```

### Introduction to DataFrames
Creating 2D tabular data structures from dictionaries.


```python
##Dataframe
## Create a dataframe from a dictionary of list 

data={
    'Name': ['Anith','Ram','Ravi'],
    'Age' : [20,22,19],
    'City': ['Chennai','Madurai','Dindigul']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

### Interoperability with NumPy
Converting DataFrames to NumPy arrays for numerical processing.


```python
import numpy as np
np.array(df)
```

### DataFrames from List of Dictionaries
An alternative construction method common in API data handling.


```python
##create a datafram from a list of dictionary 

data=[
    {'Name':'Anith','Age':20,'city':'chennai'},
    {'Name':'ram','Age':22,'city':'kovi'},
    {'Name':'Ravi','Age':18,'city':'chennai'}
]

df=pd.DataFrame(data)
print(df)
print(type(df))
```

### Importing Seaborn
Loading the library for statistical data visualization.


```python
import seaborn as sns
```

### Loading Sample Dataset
Utilizing built-in datasets like tips for practice.


```python
##basic plotting with seaborn

tips=sns.load_dataset('tips')
tips
```

### Creating a Scatter Plot
Visualizing the relationship between total bill and tip amounts.


```python
##create a scatter plot

import matplotlib.pyplot as plt 

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter Plot")
plt.show()
```

### Line Plotting
Showing trends in bill amounts based on size.


```python
##line plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot")
plt.show()
```

### Categorical Bar Plot
Comparing bill totals across different days of the week.


```python
##categorical plot
##bar plot 

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("bar Plot")
plt.show()
```

### Statistical Box Plot
Analyzing the distribution of bills per day to identify outliers.


```python
##box plot 

sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("Box Plot")
plt.show()
```

```python
##violin plot 

sns.violinplot(x='day',y='total_bill',data=tips)
plt.title("Violin Plot")
plt.show()
```

```python
##Histogram plot
sns.histplot(tips['total_bill'],bins=10,kde=False)
plt.title("Histogram Plot")
plt.show()
```

```python
##Histogram plot
sns.histplot(tips['total_bill'],bins=10,kde=True)
plt.title("Histogram Plot")
plt.show()
```

```python
##KGE plot
sns.kdeplot(tips['total_bill'],shade=True)
```

```python
##Pair Plot
sns.pairplot(tips)
```

```python
##heat map

corr=tips[['total_bill','tip','size']].corr()
corr

sns.heatmap(corr,annot=True,cmap='coolwarm')
```

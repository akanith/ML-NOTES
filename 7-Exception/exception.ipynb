### Basic Exception Handling
Using try and except to catch runtime errors.


```python
## Exception try, except block

try:
    a=b
except:
    print("This varibel is not assigned")
```

### Catching Specific Exceptions
Identifying the error type (e.g., NameError) and accessing the error message.


```python
try:
    a=b
except NameError as ex:
    print(ex)
```

### Handling Multiple Exceptions
Managing distinct error cases like ZeroDivisionError separately from a general Exception.


```python
try:
    result =1/2
    a=b 
except ZeroDivisionError as ex:
    print(ex)
    print("Please Enter the denominator Greate then 0")

except Exception as ex1:
    print(ex1)
    print("Main Exection got caught here")
```

### Try, Except, Else, and Finally
The full set of blocks for robust error handling and resource cleanup.


```python
try:
    num=int(input("enter a number "))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter the denominator greate then 0")

except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")

finally:
    print("This is the finaly block")
    print(result)
```

### Exception Handling in File Operations
Ensuring files are closed properly even if an error occurs during processing.


```python
## file handling and Exception handling 

try:
    file=open('example1.txt','r')
    content =file.read()
    print(content)

except FileNotFoundError:
    print("This file does not exists")

finally:
    if 'file' in locals() and not file.closed():
        file.close()
        print('file close')
```

### Concept of Polymorphism
Different classes having methods with the same name, providing a uniform interface.


```python
## BAse class
class Animal:
    def speak(self):
        return "Sound of the animal" 

##Derived class 1
class Dog(Animal):
    def speak(self):
        return "Woof!" 

##Derived class 2
class Cat(Animal):
    def speak(self):
        return "Meow"

## Fuction that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())

dog=Dog()
cat=Cat()
print(dog.speak())
print(cat.speak())
animal_speak(dog)
```

### Polymorphism with Shapes
Calculating area for different shapes using a shared method name.


```python
### Polymorphism with function and methods

##base class 
class Shape:
    def area(self):
        return "The area of the figure"

##derived class 1
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height

##derivede class 2
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        return 3.14*self.radius*self.radius

##Function that demonstrates polymorphism

def print_area(Shape):
    print(f"the are is {Shape.area()}")

rectangle=Rectangle(4,5)
circle=Circle(3)

print_area(rectangle)
print_area(circle)
```

### Polymorphism with Abstract Base Classes (ABC)
Enforcing method implementation in derived classes using the abc module.


```python
## Polymorphism with Abstract base classes

from abc import ABC,abstractmethod

## Define an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engin(self):
        pass

##Derived class 1
class Car(Vehicle):
    def start_engin(self):
        return "Car engine Started"

##Derived class 1
class Motorcycle(Vehicle):
    def start_engin(self):
        return "Motorcycle engine Started"

##Function that demonstrated polymorphism
def start_vechile(Vehicle):
    print(Vehicle.start_engin())

##Create object of car and motorcycle

car=Car()
motorcycle=Motorcycle()

start_vechile(car)
start_vechile(motorcycle)
```

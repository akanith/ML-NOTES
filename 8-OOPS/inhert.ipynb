### Single Inheritance: Parent Class
Defining a base class Car with shared properties.


```python
## Inheritance 
## Parent Class

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype}")
        
```

### Instantiating the Parent Class


```python
car1=Car(4,5,"Petrol")
car1.drive()
```

### Single Inheritance: Child Class
Creating a Tesla class that inherits from Car and adds specific features.


```python
class Tesla(Car):
    def __init__ (self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving = is_selfdriving
    
    def selfdriving(self):
        print("Tesla supports self driving  : {self.is_selfdriving} car")
```

### Instantiating the Child Class


```python
tesla1=Tesla(4,5,"electrice",True)

tesla1.selfdriving()
```

### Accessing Inherited Methods
The Tesla object can use methods defined in the Car class.


```python
tesla1.drive()
```

### Multiple Inheritance
Demonstrating how a class can inherit from multiple base classes (Animal and Pet).


```python
## Multiple Inheritance
## When a class inherits from more than one base class.

##Base Class 1
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print("Subclass must implement this method")


##Base Class 2
class Pet:
    def __init__(self,owner):
        self.owner=  owner

##Dervied class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        return f"{self.name} say woof"

##create an object

dog = Dog("Tommy","Krish")

print(dog.speak())
print(f"Owner:{dog.owner}")
```

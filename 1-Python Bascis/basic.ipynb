Variables in Python are case-sensitive.


```python
#Case sensitivity
name = "Python"
Name="New Py"
print(name)
print(Name)
```

```python
#Indentation

age=32
if age>=20:
    print("eligable",age)
else:
    print("Not Eligible",age)
```

### Line Continuation
Using backslash \ for multi-line statements.


```python
##list continuation
total=1+2+3+4+5+6+\
    4+5+6
print(total)
```

### Multiple Statements on a Single Line

```python
##Multiple statements on a single line
x=5;y=10;z=x+y
print(z)
```

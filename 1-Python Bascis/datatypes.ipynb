### Integer Data Type
Integers are whole numbers without a decimal point.


```python
 ##Integer Example

age=35
type(age)
```

### Floating Point Data Type
Floating point numbers have decimal places.


```python
##Floating point datatype
height=5.11
print(height)
print(type(height))
```

### Arithmetic Operators
Arithmetic operators are used to perform mathematical operations.


```python
##Arithmethic operation

a=10
b=11

add_result=a+b
sub_result=a-b
mul_result=a*b
div_result=a/b
floor_div_result=a//b
mod_result=a%b
exp_result=a**b

print(add_result)
print(sub_result)
print(mul_result)
print(div_result)
print(floor_div_result)
print(mod_result)
print(exp_result)
```

### Comparison Operators: Greater Than

### Logical Operators: AND
Returns True if both statements are true.


```python
## AND

x=True
y=True

result= x and y
print(result)


x=True
y=False

result= x and y
print(result)

x=False
y=False

result= x and y
print(result)

x=False
y=True
result= x and y
print(result)
```

### Logical Operators: OR
Returns True if at least one of the statements is true.


```python
## OR

x=True
y=True

result= x or y
print(result)


x=True
y=False

result= x or y
print(result)

x=False
y=False

result= x or y
print(result)

x=False
y=True
result= x or y
print(result)
```

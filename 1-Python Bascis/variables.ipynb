```python
## Understanding Variable types
## Python is dynamical type ,type of variable is determined at runtime

age=25 #int
height=5.7 #float
name ="Anith"#String
is_student=True#boolean

print(type(height))
```

### Type Conversion (Integer to String)
Converting numerical data to text format.


```python
age=25
print(type(age))

##Type conversion
age_str=str(age)
print(age_str)
print(type(age_str))
```

### JSON Serialization and Parsing
Converting dictionaries to JSON strings and back.


```python
 ## Data Serialization

import json
data ={'name':'Anith','age':23}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parse_data=json.loads(json_str)
print(parse_data)
print(type(parse_data))
```

### Reading and Writing CSV Files
Using the csv module to handle tabular data.


```python
## CSV

import csv
with open('example.csv',mode='w',newline="") as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['anith',20])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

### Date and Time Manipulation
Using datetime and timedelta to calculate durations and manage timestamps.


```python
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

### Regular Expressions
Finding patterns in strings using the re module.


```python
## Regular expression
import re

pattern =r'\d+'
text='there are 123 apples 456'
match=re.search(pattern,text)
print(match.group())
```

### Standard Library Import
Using import to bring in functionality from the Python Standard Library.


```python
import math
math.sqrt(16)
```

### Selective Module Import
Importing specific functions from a module.


```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

### Importing with Aliases
Using as to provide a shorter name for a library.


```python
import numpy as np 
np.array([1,2,3,4])
```

### Wildcard Import
Importing all functions from a module (generally discouraged).


```python
from math import *
print(sqrt(100))
print(pi)
```

### Importing from Local Packages
Accessing code from custom project modules.


```python
from ImportPackages.math import addition
addition(5,10)
```

### Reading a Whole File
Using with open() to read the entire content of a text file.


```python
## read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

### Reading File Line by Line
Iterating through the file object and stripping whitespace.


```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())## strip() remove the newline

```

### Overwriting a File
Using w mode to replace existing content with new data.


```python
## Writing a file(overwriting)

with open('example.txt','w') as file:
    file.write('this is a new line\n')
    file.write('this is the next new line')
```

### Appending to a File
Using a mode to add content to the end without deleting existing data.


```python
## Writing a file(without overwriting)

with open('example.txt','a') as file:
    file.write('\nthis is new line')
    file.write("\nthis is a appened line\n ")
```

### Writing Multiple Lines
Using writelines() to save a list of strings to a file.


```python
###Writing a list of line to a file

lines=['First line \n','Second line \n','Third line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

### Working with Binary Files
Writing raw bytes to a file using wb mode.


```python
### Binary files

data =b'\x00\x01\x02'
with open('example.txt','wb') as file:
    file.write(data)
```

### Reading Binary Files
Retrieving raw bytes from a file using rb mode.


```python
## Reading a binary file 
with open('example.txt','rb') as file:
    content=file.read()
    print(content)
```

### Copying File Content
Loading content from one file and saving it into another.


```python
#### Read the content form source text file and write to a destination text file 
## copying a text file

with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

### Read and Write Mode (w+)
Opening a file for both reading and writing, performing seek operations.


```python
### w+ mode where we can wirte and read in the same file and if that is not exist it will create new file 


with open('example.txt','w+') as file:
    file.write("Hello World\n")
    file.write("This is a new Line \n")

    ## Move the file cursor to the begining 
    file.seek(0)

    ## Read the content of the file

    content=file.read()
    print(content)
```

### Creating New Directories
Using os.mkdir() to manage the file system structure.


```python
### create new directory
import os
new_dir="package"

os.mkdir(new_dir)
print(f"Directory '{new_dir}' create")

```

### Listing Items in a Directory
Using os.listdir() to see files and folders.


```python
### Listing file an directory 
items=os.listdir('.')

print(items)
```

### Joining Paths
Using os.path.join() for cross-platform compatible path construction.


```python
dir_name="folder"
file_name="file.txt"
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

### Checking Path Existence
Validating if a file or directory exists before operating on it.


```python
path='example1.txt'
if os.path.exists(path):
    print(f"The path '{path}' exist")
else:
    print(f"the path '{path} doew not exists")
```

### Getting Absolute Paths
Resolving relative paths to their full system location.


```python
## getting the absolute path
import os
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```

### Iterables and Loops
Basics of iterating over a list using a for loop.


```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

### Checking the Type


```python
type(my_list)
```

### Printing the Iterable


```python
print(my_list)
```

### Creating an Iterator Object
Using iter() to transform an iterable into an iterator.


```python
## Iterator
iterator=iter(my_list)
print(type(iterator))
```

### Using next() to Retrieve Elements
Manual iteration using the next() function.


```python
##Iterate thorug all the element

next(iterator)
```

### Re-initializing the Iterator


```python
iterator=iter(my_list)
```

### Handling StopIteration Exception
Gracefully catching the end of an iterator.


```python
try:
    print(next(iterator))
except StopIteration:
    print("There are no elements in the iterator")
```

### String Iterators
Strings are also iterables; you can create iterators for them.


```python
## String iterator 

my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator)) #Output:H
print(next(string_iterator)) #Output:e
```

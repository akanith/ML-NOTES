### Function as an Object
Assigning a function to a new variable and proving it persists even if the original name is deleted.


```python
## function copy
def welcome():
    return "welcom to the advnaced python course"

welcome()
```

### Closures: Nested Functions
Passing parameters from an outer function to an inner function.


```python
wel=welcome
print(wel())
del welcome
print(wel())
```

### Calling the Closure Function


```python
## Closures


def main_welcome(msg):
    def sub_welcome_method():
        print("Welcome to the advanced pyton course")
        print(msg)
        print("Learn this concepts properly")
    return sub_welcome_method()
```

### Passing Functions as Arguments
A fundamental prerequisite for decorators.


```python
main_welcome("Welcom Every one")
```

### Implementation of a Higher-Order Function


```python
def main_welcome(func,list):
    def sub_welcome_method():
        print("Welcome to the advanced pyton course")
        print(func(list))
        print("Learn this concepts properly")
    return sub_welcome_method()
```

### Step-by-Step Decorator Concept
Wrapping a functions execution with additional logic manually.


```python
main_welcome(len,[1,2,3,4])
```

### Standard Function Definition


```python
##Decorator

def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advanced pyton course")
        func()
        print("Learn this concepts properly")
    return sub_welcome_method()
```

### Decorating Manually
Passing one function into another to extend its behavior.


```python
def course_intro():
    print("this is an advance python class")

course_intro()
```

### Simplified Syntax: The @ Decorator
Using the @ symbol to apply decorators automatically.


```python
main_welcome(course_intro)
```

### Wrapper Function Pattern
The standard implementation of a decorator using a wrapper function.


```python
## we will be using '@' for decorators


@main_welcome
def course_intro():
    print("this is an advance python class")
```

### Decorating with @ Syntax


```python
##Decorator

def my_method(func):
    def wrapper():
        print("This part print befroe Function")
        func()
        print("This part print after Function")
    return wrapper()
```

### Advanced: Decorators with Arguments
Using three levels of nested functions to pass parameters to the decorator itself.


```python
@my_method
def say_hello():
    print("Hello")
```

### Applying a Parameterized Decorator


```python
## Decorators with arguments

def repeat(n):
    def decorator(fun):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                fun(*args,**kwargs)
        return wrapper
    return decorator
```

### Executing the Decorator logic


```python
@repeat(3)

def say_hello():
    print("Hello")
```

```python
say_hello()
```

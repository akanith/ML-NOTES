### For Loop
Basic iteration over a range of numbers.


```python
##for loop

for i in range(1,5):
    print(i)
    
```

### For Loop with Step
Using the start, stop, and step arguments in range().


```python
for i in range(1,10,2):
    print(i)
```

### For Loop in Reverse
Iterating backwards using a negative step.


```python
for i in range(10,1,-2):
    print(i)
```

### Iterating Over Strings
Each character in the string is processed in the loop.


```python
## string 

str="Anith"

for i in str:
    print(i)
    
```

### While Loop
Executing a block as long as a condition remains true.


```python
## While loop

count =0
while count<5:
    print(count)
    count=count+1
    
```

### While Loop with Logic
Using conditions within the loop body.


```python
count=0
while count%2==0:
    print(count)
    count=count+1
```

### Loop Control: Break
Exiting the loop immediately when a condition is met.


```python
## Loop control statement 

for i in range(10):
    if i==5:
        break
    print(i)
```

### Loop Control: Continue
Skipping the rest of the current iteration.


```python
## Continue

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

### Loop Control: Pass
A placeholder statement that does nothing.


```python
## Pass
##it is a null operation,it does nothing 

for i in range(5):
    if i==3:
        pass
    print(i)
```

### Nested Loops
Executing a loop inside another loop.


```python
##nested loops
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
        
```

### Practical Example: Sum of Numbers (While)
Calculating sum using a while loop.


```python
n=10
sum=0
count =1

while count<=n:
    sum =sum+count
    count =count +1

print("Sum of first 10 number = ",sum)
```

### Practical Example: Sum of Numbers (For)
Calculating sum using a for loop.


```python
n=10
sum=0
for i in range(11):
    sum=sum+i
print(sum)
```

### Prime Number Search
A complex example using nested loops and else with loops.


```python
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

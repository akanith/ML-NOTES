### Simple If Statement
Executing code only if a condition is true.


```python
## if statement

age=18
if age>=18:
    print("Allowed to Vote")
```

### If-Else Statement
Handling both true and false outcomes.


```python
## else

age=15
if age>=18:
    print("Allowed to Vote")
else:
    print("Not Allowed to Vote")
```

### Elif (Else-If) Statement
Checking multiple distinct conditions.


```python
## elif

age=12

if age<13:
    print("your child")
elif age<18:
    print("your teenager")
else:
    print("your adult")
```

### Nested Conditionals
Using an if statement inside another if statement.


```python
#nested condition 


num=int(input("Enter a number"))

if num>=0:
    print("Postivie")
    if num%2==0:
        print("Even")
    else:
        print("Odd")
else:
    print("Negative")
```

### Practical Example: Leap Year
A common logic puzzle solved with nested conditions.


```python
## practical example

year=int(input("Enter a Year"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is a leap year")
    else:
        print(year,"is a leap year")
else:
    print(year,"not a leap year")
```

### Introduction to Dictionaries
Creating an empty dictionary.


```python
## Creating a dictionary

empty_dict={}
print(type(empty_dict))
```

### Key-Value Pairs
Defining a dictionary with initial data.


```python
student={"name":"John","age":25,"gender":"Male"}

print(student)
print(type(student))
```

### Accessing Elements
Using keys and the get() method.


```python
## accessing dictionary elements

student={"name":"John","age":25,"gender":"Male","Grade":"A"}

print(student)

print(student["name"])
print(student["age"])
print(student["gender"])

## Accessing using get() method
print(student.get('Grade'))
print(student.get('last_name',"Not Available"))
```

### Mutability of Dictionaries
Dictionaries allow adding, updating, and deleting items.


```python
## Modifying Dicitonary Elements
##This is Mutable,so you can add,update or delete elements

print(student)
```

### Modifying and Adding Items
Updating existing values or creating new key-value pairs.


```python
student["age"]=33
print(student)
student["address"]="India"
print(student)
```

### Deleting Items
Using del to remove entries.


```python
del student['Grade']

print(student)
```

### Dictionary Methods
Accessing keys, values, and full items.


```python
## Dictonary Methods
keys=student.keys()
print(keys)
values=student.values()
print(values)

items= student.items()
print(items)
```

### Copying Dictionaries: Reference Copy
Changes to the original affect the copy.


```python
## Shallow Copy
student_copy=student
print(student)
print(student_copy)
```

### Illustrating Reference Copy behavior


```python
student["name"]="Raj"
print(student)
print(student_copy)
```

### Copying Dictionaries: Shallow Copy
Using .copy() to create an independent object.


```python
student_copy1=student.copy()##shallow copy
print(student_copy1)
print(student)
```

### Illustrating Shallow Copy behavior


```python
student["name"]="Ravi"
print(student_copy1)
print(student)
```

### Iterating Over Keys
Standard loop through dictionary keys.


```python
## Iteration over dictionaries

for keys in student.keys():
    print(keys)
```

### Iterating Over Values
Using the .values() method.


```python
## Iteration over value

for values in student.values():
    print(values)
```

### Iterating Over Items
Using .items() to get both key and value.


```python
## Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

### Nested Dictionaries
Storing dictionaries within dictionaries.


```python
## Nested Dictionaries

students={
    "student1":{"name":"Ram","age":34},
    "student2":{"name":"Ravi","age":24}
    }
print(students)
```

### Accessing Nested Data


```python
## Acces nested Dictionaries elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

### Iterating Over Nested Dictionaries


```python
## iterating over ested dictionaries

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")

print("outside the loop")
```

### Dictionary Comprehension
Efficiently creating dictionaries using expressions.


```python
## Dictionary comphrehension

squares={x:x**2 for x in range(5)}
print(squares)
```

### Conditional Dictionary Comprehension


```python
## condition dictionary comphrehension

squares={x:x**2 for x in range(11) if x%2==0}
print(squares)
```

### Practical Example: Frequency Counter
Counting occurrences of items in a list.


```python
## Practical example

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1

print(frequency)
```

### Merging Dictionaries
Using unpacking operators to combine two dictionaries.


```python
## Merge 2 dictionaries into one 

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```

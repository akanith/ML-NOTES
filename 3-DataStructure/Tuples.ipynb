### Introduction to Tuples
Tuples are immutable sequences, defined using parentheses.


```python
## create a tuple 

empty_tuple =()
print(empty_tuple)
print(type(empty_tuple))
```

### Tuple Constructor
Creating lists and tuples using constructors.


```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

### Tuples from Lists
Converting other iterables into tuples.


```python
number=tuple([1,2,3,4,5,6])
number
```

### Accessing Tuple Elements
Indexing works similarly to lists, but values cannot be changed.


```python
## Accessing Tuple elements

print(number[2])
print(number[-1])
```

### Tuple Methods
Counting occurrences and finding indices.


```python
## Tuple Methods

print(number.count(1))
print(number.index(3))
```

### Packing and Unpacking Tuples


```python
## Packing and Unpacking Tuple 

packed_tuple=1,"Hello",3.14
print(packed_tuple)
```

### Unpacking Values
Assigning tuple elements to multiple variables.


```python
## Unpacked Tuple 

a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

### Extended Unpacking
Using the asterisk * to capture multiple elements.


```python
## Unpacked with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

### Nested Tuples
Accessing elements within tuples that contain other tuples.


```python
##  Nested Tuple 


nested_tuple =((1,2,3),("a","b","c"),(True,False))

print(nested_tuple[0])
print(nested_tuple[1][2])
```

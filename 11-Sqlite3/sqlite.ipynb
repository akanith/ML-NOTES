### Database Connection
Establishing a connection to a SQLite database file.


```python
import sqlite3

# This will work without any installation
conn = sqlite3.connect('example.db')
conn
```

### Creating a Cursor Object
The cursor acts as an interface to execute SQL commands.


```python
cursor=conn.cursor()
```

### Defining the Schema
Creating a table structure with appropriate data types.


```python
##create table 

cursor.execute('''
create table if not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age int,
    dep text
    )
''')

##Commit the change 

conn.commit()
```

### Initial Data Verification
Running a select query to ensure the table is ready.


```python
cursor.execute('''
Select * from employees
''')
```

### Inserting Records
Adding multiple rows of data to the table and committing the transaction.


```python
##insert the data in sqlite table

cursor.execute('''
Insert Into employees(name,age,dep)
values("Anith",20,"CSE")
''')

cursor.execute('''
Insert Into employees(name,age,dep)
values("Ram",20,"CSE")
''')

cursor.execute('''
Insert Into employees(name,age,dep)
values("Ravi",20,"CSE")
''')

conn.commit()
```

### Querying and Fetching Results\Retrieving stored records and displaying them.


```python
##Query the data from the table 
cursor.execute('Select * from employees')
rows=cursor.fetchall()

##print all the qure data

for row in rows:
    print(row)
```

### Updating Records
Modifying existing data using the UPDATE statement.


```python
##Update the data in the  table 

cursor.execute(''' 
UPDATE employees
set age=23
where name="Ravi"
''')

conn.commit()
```

### Verifying Progress
Executing a SELECT statement to list current rows.


```python
##Query the data from the table 
cursor.execute('Select * from employees')
rows=cursor.fetchall()

##print all the qure data

for row in rows:
    print(row)
```

### Working with Sales Data
Connecting to a separate database file for a different business context.


```python
##Working with sales data
##connect to an Sqlite database

connection =sqlite3.connect('sales_data.db')
cursor=connection.cursor()

##create table

cursor.execute('''
    CREATE TABLE IF NOT EXISTS sales(
        id int Primary key,
        date text not null,
        product text not null,
        sales int,
        region text
    )
''')
```

### Efficient Batch Inserts
Using executemany() to handle large lists of data efficiently.


```python
##Insert data into the sales table 

sales_data=[
    ('1-2-2020','apple',100,'north'),
    ('1-2-2020','banana',50,'south'),
    ('1-2-2020','graps',90,'east'),
    ('1-2-2020','mango',160,'west'),
    ('1-2-2020','jackfruit',200,'north')
]
cursor.executemany('''
INSERT into sales(date,product,sales,region)
    values(?,?,?,?)
''',sales_data)

connection.commit()
```

### Verifying Progress
Executing a SELECT statement to list current rows.


```python
##Query the data from the table 
cursor.execute('Select * from sales')
rows=cursor.fetchall()

##print all the qure data

for row in rows:
    print(row)
```

### Closing the Connection
Releasing system resources by terminating the database session.


```python
##close connection
connection.close()
```

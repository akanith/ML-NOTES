# Memory Management in Python

Python manages memory automatically using a combination of **Reference Counting** and a **Garbage Collector**. Understanding these mechanisms helps in writing efficient code and avoiding memory leaks.

## 1. Reference Counting

Reference counting is the primary mechanism Python uses to track objects in memory. Every object has a count of how many references point to it. When an object's reference count drops to zero, Python's memory manager automatically deallocates it.

```python
import sys

a = []
## The reference count is 2: one from 'a' and one from the getrefcount() function itself
print(sys.getrefcount(a))
```

### Increasing Reference Count
Creating a new reference to an existing object increases its reference count.

```python
b = a
print(sys.getrefcount(a))
```

### Decreasing Reference Count
Deleting a reference or reassigning it reduces the count. Once it reaches zero, the memory is freed.

```python
del b
print(sys.getrefcount(a))
```

## 2. Garbage Collection

While reference counting handles most memory management, it cannot handle **circular references** (where two objects refer to each other). Python's Garbage Collector (GC) scans for these cycles and cleans them up.

```python
import gc

## Check if garbage collection is enabled
print(f"GC enabled: {gc.isenabled()}")

## Manually trigger garbage collection
gc.collect()
```
